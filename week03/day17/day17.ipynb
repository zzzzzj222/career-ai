# 部分变量预填充

```python
from langchain_core.prompts import ChatPromptTemplate

'''部分变量预填充'''

template = ChatPromptTemplate.from_messages([
    ("system", "你是{department}的{role}"),
    ("user", "{ask}"),
]) # 传入模板字符串，生成ChatPromptTemplate对象

it_template=template.partial(department="IT", role="技术")

sales_template=template.partial(department="销售", role="销售经理")


print(it_template.invoke({"ask":"为什么每年年底汽车会促销"}))
print(sales_template.invoke({"ask":"为什么每年年底汽车会促销"}))
```

# 消息占位符

```python
from langchain_core.prompts import ChatPromptTemplate
'''消息占位符'''
placeholder_tmp=ChatPromptTemplate.from_messages([
    ("system", "你是一个ai助手"),
    ("placeholder", "{conversation}"),
]) # 传入模板字符串，生成ChatPromptTemplate对象

res=placeholder_tmp.invoke({
    "conversation": [
        ("user", "我的快递怎么还没到？"),
        ("assistant", "订单查询"),
        ("user", "我想买一双运动鞋"),
        ("assistant", "产品推荐"),
        ("user", "耳机支持主动降噪吗？"),
        ("assistant", "商品咨询"),
    ]
})
print(res)
```

# 使用MessagePlaceholder

```python
from langchain_core.prompts import MessagesPlaceholder,ChatPromptTemplate
'''使用MessagePlaceholder'''
template = ChatPromptTemplate.from_messages([
    ("system", "你是一个ai助手"),
    MessagesPlaceholder(variable_name="conversation"),
])

res=template.invoke({
    "conversation": [
        ("user", "我的快递怎么还没到？"),
        ("assistant", "订单查询"),
        ("user", "我想买一双运动鞋"),
        ("assistant", "产品推荐"),
        ("user", "耳机支持主动降噪吗？"),
        ("assistant", "商品咨询"),
    ]
})
print(res)

```

# 存储历史消息

```python
from langchain_core.prompts import MessagesPlaceholder,ChatPromptTemplate
'''存储历史消息'''
template=ChatPromptTemplate.from_messages([
    ("system", "你是一个ai助手"),
    MessagesPlaceholder(variable_name="history"),
    ("user","{conversation}"),
]) 

template.invoke({
    "history":[
        ("user","你好"),
        ("assistant","你好！有什么我可以帮忙的吗？")
    ],
    "conversation":"我想买一双运动鞋"
}) # 会将历史消息和当前消息一起传递给模型，方便模型理解对话历史。

```
