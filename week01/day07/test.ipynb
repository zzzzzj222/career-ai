```python
import pandas as pd

# 清洗数据

df = pd.read_csv('D:\\VSProject\\career-ai\\high_rated_movies.csv')

# 去掉上映日期中的换行符和多余空白
# 例如："\n        2026-08-25 (US)\n      " -> "2026-08-25 (US)"
df['上映日期'] = df['上映日期'].fillna('').astype(str).str.replace(r'\s+', ' ', regex=True).str.strip()
df['时长'] = df['时长'].fillna('').astype(str).str.replace(r'\s+', ' ', regex=True).str.strip()
# 直接写回原文件，覆盖原 CSV
df.to_csv('D:\\VSProject\\career-ai\\high_rated_movies.csv', index=False, encoding='utf-8')

df['上映日期'].describe()

df['时长'].describe()
```
