```python
import matplotlib.pyplot as plt
import numpy as np
# 在0到2π之间均匀生成200个数，作为横坐标
x = np.linspace(0, 2 * np.pi, 200)
# 计算每个横坐标对应的正弦值，作为纵坐标
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
# 在0到2π之间均匀生成200个数，作为横坐标
x = np.linspace(0, 2 * np.pi, 20)

fig, axes = plt.subplots()
plt.title("正弦余弦图", fontsize=15)
# 设置x,y轴
plt.xlabel('时间')
plt.ylabel('高度')
# 设置全部展示
plt.xticks(x[::2])
# 显示网格,设置网格样式
plt.grid(ls='--',alpha=0.3)

axes.plot(x, np.sin(x),label='sin')
axes.plot(x, np.cos(x),label='cos')
# 在plot中指定名称后，使用legend显示
plt.legend(loc="best")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 多图表绘制（子图绘制）
fig,axe=plt.subplots(nrows=2,ncols=2,figsize=(20,6),dpi=100)
x = np.linspace(0, 2 * np.pi, 20)
axe[0,0].plot(x, np.sin(x),label='sin')
axe[1,0].plot(x, np.cos(x),label='cos')
axe[0,1].plot(x,np.random.randint(1, 51, size=20))
axe[1,1].plot(x,np.random.randint(1, 51, size=20))

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
fig, axe = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)
# x轴城市名称
x=['北京', '上海', '广州', '深圳', '杭州', '南京', '武汉', '成都', '重庆', '天津']
axe[0].bar(x, np.random.randint(1, 51, size=10),color='orange', label='城市人口')
axe[0].set_xlabel('城市')
axe[0].set_ylabel('人口')
axe[0].set_title('各城市人口分布')
axe[0].legend()

axe[1].pie(np.random.randint(1, 51, size=10), labels=x, autopct='%1.1f%%', startangle=90)
axe[1].set_title('城市人口比例')
axe[1].legend(loc='lower center',ncol=5, bbox_to_anchor=(0.5, -0.1))

# 保存图片
plt.savefig('city_population.png')

plt.show()
```
